```python
#Linear Regression using normal equation

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

class LinearRegression:

  def __init__(self):
    self.theta=None


# jab dataframe me bias column ya pd se value add karenge to original effect na ho do copy


  def fit(self,X_train,y_train):

    X_train=np.array(X_train)

    y_train=np.array(y_train)

    X_train=np.hstack((np.ones((X_train.shape[0],1)),X_train))


    self.theta=np.linalg.solve((X_train.T @ X_train),(X_train.T@ y_train))


  def weight(self):
    return self.theta[1:]

  def predict(self,X_test):

    X_test=np.asarray(X_test)
    if X_test.ndim==1:

      X_test=X_test.reshape(1,-1)
      # X_test=X_test.reshape(1,X_test.shape[0])

    X_test=np.hstack((np.ones((X_test.shape[0],1)),X_test))


    return X_test @ self.theta

  def mean_squared_error(self,X_test,y_test):

    y_test=np.asarray(y_test)
    pred=self.predict(X_test)
    return np.mean((y_test-pred)**2)

  def mean_absolute_error(self,X_test,y_test):

      y_test=np.asarray(y_test)
      pred=self.predict(X_test)
      return np.mean(np.abs(y_test-pred))

  def r2_score(self,X_test,y_test):

      y_test=np.array(y_test)
      rs=np.sum((y_test-self.predict(X_test))**2)
      rt=np.sum((y_test-np.mean(y_test))**2)
      return 1-(rs/rt)

  def score(self,X_test,y_test):
    return self.r2_score(X_test,y_test)
```

```python
# Linear Regression using gradient decent

class LinearRegGd:

  def __init__(self,learning_rate=0.01,epochs=1000):

    self.learning_rate=learning_rate
    self.epochs=epochs
    self.theta=None


  def fit(self,X_train:np.ndarray,y_train:np.ndarray):

    self.mean=np.mean(X_train,axis=0)
    self.std=np.std(X_train,axis=0)
    self.std[self.std==0]=1

    X_train=(X_train-self.mean)/self.std

    y_train=y_train.reshape(-1,1)

    X_train=np.hstack((np.ones((X_train.shape[0],1)),X_train))

    self.theta=np.zeros((X_train.shape[1],1))

    self.cost_history=[]

    for epo in range(self.epochs):

      if self.theta is None:
        raise ValueError("Model in not fit")


      pred = X_train @ self.theta

      error=pred - y_train

      grad=X_train.T @ error/X_train.shape[0]

      self.cost_history.append(np.sum(error**2)/(2*X_train.shape[0]))

      self.theta=self.theta - self.learning_rate * grad

    return self



  def predict(self,X_test:np.ndarray):

    if X_test.ndim==1:

      X_test=X_test.reshape(1,-1)
      # X_test=X_test.reshape(1,X_test.shape[0])


    X_test=(X_test-self.mean)/self.std


    X_test=np.hstack((np.ones((X_test.shape[0],1)),X_test))

    return X_test @ self.theta

  def cost_plot(self):

    plt.plot(self.cost_history)
    plt.title("epochs vs cost graph")
    plt.xlabel("epochs")
    plt.ylabel("cost")
    plt.show()

  def mean_squared_error(self,X_test,y_test):

    y_test=np.asarray(y_test).reshape(-1,1)
    pred=self.predict(X_test)

    return np.mean((y_test-pred)**2)

  def mean_absolute_error(self,X_test,y_test):

      y_test=np.asarray(y_test).reshape(-1,1)
      pred=self.predict(X_test)

      return np.mean(np.abs(y_test-pred))

  def r2_score(self,X_test,y_test):

      y_test=np.array(y_test).reshape(-1,1)
      pred=self.predict(X_test)

      rs=np.sum((y_test-pred)**2)
      rt=np.sum((y_test-np.mean(y_test))**2)

      return 1-(rs/rt)

  def score(self,X_test,y_test):

    return self.r2_score(X_test,y_test)

```
